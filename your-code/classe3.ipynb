```python
principal = 1000
rate = 5 # 5% interest
years = 10

# Calculate the compound interest using the formula A =P(1+r/n)^(nt)
# A = final amount
# P = Principal
# r= annual interest
# n = number of times interest is compounded per year
# t = time in years
```

```python
n = 1
current_year = 1

while current_year <= years:
    amount= principal*(1+(rate/(100*n)))**(n*current_year)
    print (f"Year {current_year}: ${amount:.2f}")
    current_year +=1
```

```python
sales_data = [ 200, 300, 400 , 500, 600, 700, 800]

#Initialize total_sales to 0 BEFORE the loop
total_sales = 0

#Initialize max_sales before startijng the loop
max_sales = 0
```

### Use if statemets to calculate wich day had the maximum sales 

```python
for day in sales_data:
    total_sales += day
    if day > max_sales:
        max_sales = day

print(total_sales)
print(max_sales)
```

## Nested List Comprehensions

```python
nested_list = [[1,2],[3,4],[5,6]]
# create a flat list of individual ints
```

```python
new_list = []
for i in nested_list:
    for k in i:
        new_list.append(k)
new_list
```

```python
flattened_list = [k for i in nested_list for k in i]
flattened_list
```

```python
flattened_list_squared = [k**2 for i in nested_list for k in i]
flattened_list_squared
```

```python
nested_list = [[[[1,2], [3,4]]]]
```

```python
new_list = []
for i in nested_list:
    for k in i:
        for j in k:
            for h in j:
                new_list.append(h)
new_list
```

```python
new_list = [h for i in nested_list for k in i for j in k for h in j]
new_list
```

# Challenge #n
Write a dictionary containing the length of each word

```python
cities_list= ["Barcelona","Madrid", "Girona", "Murcia"] 
```

```python
cities_lengths = {}
```

```python
lengths = []
for city in cities_list:
    lengths.append(len(city))
for name,length in zip(cities_list, lengths):
    cities_lengths[name] = length
```

```python
lengths
```

```python
cities_lengths
```
